SINGLY LINKED LIST

```python
class Node:
    def __init__(self, new_data):
        self.data = new_data
        self.next = None

head = Node(10)
head.next = Node(20)
head.next.next = Node(40)
temp = head
while temp is not None:
    print(temp.data, end = " ")
    temp = temp.next
```

DOUBLY LINKED LIST

```python
class Node:
    def __init__(self, new_data):
        self.data = new_data
        self.next = None
        self.prev = None

head = Node(10)
head.next = Node(20)
head.next.prev = head

head.next.next = Node(30)
head.next.next.prev = head.next

head.next.next.next = Node(40)
head.next.next.next.prev = head.next.next

temp = head
while temp is not None:
    print(temp.data, end = " ")
    if temp.next is not None:
        print("<->", end = " ")
    temp = temp.next

```

COUNTING LINKED LIST

```python
class Node:
    def __init__(self, new_data):
        self.data = new_data
        self.next = None

def count_nodes(head):
    count = 0
    curr = head
    while curr is not None:
        count +=1
        curr = curr.next
    return count

if __name__ == "__main__":
    head = Node(10)
    head.next = Node(20)
    head.next.next = Node(30)
    head.next.next.next = Node(40)

    print("Number of nodes in the linked list:", count_nodes(head))
```
